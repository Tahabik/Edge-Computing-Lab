## Estimation.

So it's better to start with a question:<br>
Why do we use **Estimation** and what is the relation between learning and statistics?

+ Target model in the learning problems can be considered as a statistical model.
+ For a fixed set of data and underlying target (statistical model), the estimation methods try to estimate the target from the available data.

> We need estimation to estimate parameters of a distribution.

Now the purpose mainly is estimating the **probability density** function $p(x)$, given a set of data points $\{x^{(i)}\}_{i=1}^{N}$

For getting to the distribution there are two methods:

+ **Parametric**: We consider a fixed parametric model for distribution and we only try to estimate it's parameters.
    + A number of parameters are optmized by fitting the model to the data set

+ **Non-parametric**: From the start we don't consider any model on the distribution and we try to obtain or get the distribution using the data only.
    + The form of the density function is determined entirely by the data.

### Parametric density estimation:

This estimation's goal is to estimate parameters from a dataset $D = \{x^{(1)}, ..., x^{(N)}\}$.

+ D contains N independent, identically distributed training samples.<br>

We need to determine $\theta$(parameters of the distribution) given $ \{x^{(1)}, ..., x^{(N)}\}$


## Likelihood

Likelihood is the **conditional probability** of observations $D = \{x^{(1)}, x^{(2)}, ...,x^{(N)}\}$ given the value of parameters $\theta$.<br>

**Likelihood focuses on the parameters, given observed data**:

When we have a set of observed data points, the *likelihood* function asks: "How 'likely' are *these* specific parameter values to have generated the observed data?"

The likelihood function, often denoted $L(\theta | data)$ or $L(\theta;data)$, is essentially the probability (or probability density) of observing our **actual data**, but viewed as a function of the *unknown* parameters $\theta$, while keeping the data fixed.

$$L(\theta | D) = P(D|\theta) = \prod_{i=1}^{N} P(x^{(i)}|\theta)$$

+ $P(x^{(i)}|\theta)$: Represents the probability(or probability density) of observing that specific data point $x^{(i)}$, given the specific set of parameters $\theta$ for the assumed distribution.

+ **Probability density**: is a measure of the **concentration of probability** around a specific value for a **continuous random variable.** It's not a direct probability itself, but rather "probability per unit of something"(e.g, probability per unit of length, per unit of time, etc). It is the value on the **y-axis** of a PDF, it tells us where the value of the random variable are more **likely** to cluster.

### Maximum Likelihood Estimation(MLE):

Now our variable was our distribution's parameters (we know the type of distribtuion), so technically we should look for the parameters(in the above example the mean of the normal distribution.) that would make the likelihood of our samples maximum:

$$\theta_{ML} = \argmax_{\theta} P(D | \theta)$$

> The distribution that will make the production of P's the maximum value is the distribution that we're looking for.

<br>


<center>Figure 4: Adopted from [1]</center>
<br>


> The figure above is showing that the $\theta$ itself has a function with a optimal point with the result of the $P(D|\theta)$ on the plot's y-axis.

The distributions that we mostly deal with are exponential and normally we take a $\log$ from them, cause the maximum in the actual function is the same as the equal in the logarithm of that same function, So this makes the calculations easier for us:

$$L(\theta) = \ln{P(D | \theta)} = \ln{\prod_{i = 1}^{N} P(X^{(i)} | \theta)} = \sum_{i=1}^{N} \ln{P(x^{(i)} | \theta)} \\ \hat{\theta}_{ML} = \argmax_{\theta}L(\theta) = \argmax_{\theta} \sum_{i = 1}^{N} \ln{p(x^{(i)} | \theta)} $$


Thus for finding the optimal parameter we should take the gradient of the equation above and equal i t to zero:

$$\nabla_{\theta}L(\theta) = 0$$


**What is the problem with likelihood?**

If we have small number of samples the resulted estimation can be wrong and can fit so much to data and result in overfit (It will give good results for previuos fixed, known samples but if a new sample comes along and we do not know anything it can result in a wrong mapping) For solving this problem we use **MAP(Maximum A Posterior)**

### Maximum A Posterior(MAP):

We use MAP here for solving the errors that we may face due to small datasets.<br>
First Let's look at the below explanations:

$$\text{likelihood} = P(D | \theta), \; \; D = \{x^{(1)}, ..., x^{(N)}\}$$
$$\text{Prior} = P(\theta) \Rightarrow \text{Initial/prior probability of parameters}$$
$$\text{Ppsterior} = P(\theta | D) \Rightarrow \text{parameter's posterior probabilities after seeing the data}$$


Now let's explain each notation aobve a bit more:

+ **$\text{Prior} = P(\theta)$**: Thi is our belief about the parameter **before** seeing any data.

    **How can a parameter have a probability**?<br>

    In **Bayesian** statistics, we treat parameters as **random variables**. We assign a probability distribution to them, **not because they're truly random**, but because we are **modeling our uncertainaty** about them.<br>
    It is like, we don't know exact value of $\theta$ and we're modeling our uncertainaty with a distribution.

+ **$\text{Posteriro} = P(\theta | D)$**: Thi is our belief about the parameter **after** seeing the data.

    First of all look at the exact equation of posterior:
    $$P(\theta | D) = \frac{P(D | \theta).P(\theta)}{P(D)}$$

    > The above equation is saying that **observing data updates our belief** about which values of $\theta$ are more or less likely.

    **How does seeing the data change the probability distribution of a parameter?**

    + $P(\theta)$: our initial belief. A distribution of how plausible each $\theta$ is before seeing any data.
    + $P(D ∣ \theta)$: the likelihood. How likely would the data D be if $\theta$ were the true value? This is the “evidence strength” each candidate $\theta$ receives from the observed data.
    >  It tells us: "If I assume this value of $\theta$, how well does it explain what I just observed?"
    + Multiply them together: we are combining your prior belief with how well each θ explains the data.
    + $P(D)$: It is just a normalizationo costant


Now we in MLE we were looking for the $\theta$ that would make the $P(D ∣ \theta)$ maximum, but in MAP we're looking for the $\theta$ that would maximize the $P(\theta | D)$, We wanna look for a $\theta$ that maximize the probability of sampels after seeing parameters:

$$\hat{\theta}_{MAP} = \argmax_{\theta} P(\theta | D)$$


So by seeing the formula of $P(\theta | D)$ we can see that it's denominator has nothing to do with $\theta$, so we ony need to work with $P(D | \theta).P(\theta)$
