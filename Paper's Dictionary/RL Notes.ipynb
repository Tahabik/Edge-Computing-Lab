### On-policy and Off-policy

In RL, "on-policy" and "off-poliyc" refer to **how** an agent learns its **optimal policy** while interacting with its environment. The distinction lies in the relationship between the policy and the agent uses to *explore* the environment (The "behavior policy") and the policy it aims to improve or evaluate (the "target policy")

**On-policy Algorithms**:

An on-policy algorithms learns or evaluates the policy that it is currently using to interact with the environment. This means the "behavior polic" and the "target policy" are same.<br>
The agent takes actions based on its current policy, observes the outcomes (states and rewards), and then updates that same *policy* based on those experience.

On-policy methods typically use "soft" policies, meaning they still allow for some exploration (e.g., $\epsilon$-greedy policies, where the agent mostly exploits its current knowledge but occasionally takes random actions to explore). The explorationis built into the the policy itself. Like SARSA.


**Off-policy Algorithms**:

An off-policy algorithm learns or evaluates a policy that is **different** from the policy it uses to generate data or interact with the environment. The "behavior policy" is used for exploration, while the "target policy" is the one being optimized.M<br>
The agent can explore the environment using the one policy(e.g., a highly exploratory policy to gather diverse experiences) but then use these experiences to learn or improve *different*, typically  more optimal and less exploratory, policy.

This separation allows for more flexible exploration strategies. The behavior policy can be very exploratory (e.g., truly random actions.), ensuring a wider range of experiences are gathered, while the target policy can focus purely on exploition the best-known actions

## SARSA

State-Action-Reward-Action is an **on-policy** RL algorithm. This means that it updates and adjusts the *same* policy it uses for making decisions.

How it works:
1. **Observe current state(S)**: The agent starts by observing its current state in the environment.
2. **Choose Action(A)**: Based on its current policy (which includes some exploration, like $\epsilon$-greedy strategy), the agent selects an action. 
3. **Execute Actions**: The agent performs the chosen action in the environment.
4. **Receive Reward(R)**: After executing the action, the agent receives an immediate reward from the environment.
5. **Observe next state(S')**: The environment transitions to a new state as a result of the action.
6. **Choose next Action(A')**: Crucially, SARSA then choose the *next action* from the *new* state (S') using the **same current policy** that it is trying to learn and improve.
7. **Update Q-value**: The Q-value for the original state-action pair ($Q(S, A)$) is updated using the observed reward and the Q-value of the next state-action pair ($Q(S', A')$) that was just chosen, This is the core update step:

$$Q(S, A) \leftarrow Q(S, A) + \alpha[R + \lambda Q(S', A') - Q(A, A)]$$

The agent learns from its experinces by continuously updating the Q-values based on the $(S, A, R, S', A')$ tuple, were both current and next actions are chosen by the *same* evolving policy.

## Fuzzy Reinforcement Learning (FRL)

FRL combines RL with Fuzzy Logic(FL):

+ **Fuzzy Logic**: Unlike traditional "crisp" logic (where something is either true or false, 0 or 1), fuzzy logic deal with degrees of truth. It allows for reasoning with imprecise or uncertain information, using linguistic terms like "slightly warm", "very cold", "high load", or "low latency"

**How FRL works?**

FRL Integrates fuzzy logic into the component of an RL system, such as: 
+ **Fuzzy States**: Instead of discrete, exact state, states can be defined using fuzzy sets (e.g., "CPU utilization is medium" rather than "CPU utilization is exactly 75%"). This helps manage continuous or very large state spaces more intuitively.
+ **Fuzzy Actions**: Actions can also be fuzzy (e.g., "increase power slightly," "reduce offloading significantly"). 
+ **Fuzzy Rewards**: Rewards might be described fuzzily (e.g., "very good reward," "poor reward").
+ **Fuzzy Rules**: Fuzzy inference systems can be used to derive policies, often expressed as "IF-THEN" rules (e.g., "IF CPU load is high AND network latency is high THEN offload aggressively"). The RL component then learns to optimize the parameters of these fuzzy rules or the membership functions.

**Why it's used?**

FRL is particularly useful in environments that are **complex, uncertain, or have continuous variables** where it's difficult to define precise states and actions, or where expert knowledge can be incorporated. It helps in dealing with the "curse of dimensionality" to some extent by generalizing over states.
