DRL is a subclass of RL methods, that overcomes the constraints of RL methods in continuous and big state action environemnts that they cannot use tabular representation and storing for state-action table and values, istead we use a DNN(Deep Neural Network) to overcome the reinforcement learning: $Q(s, a, \theta) \approx Q_{\pi}(s, a)$. 

First of all let's see what were "tabular representations":

In traditional or basic RL algorithms like Q-learning or SARSA, "tabular representations" refer to literally **tables** (like spreadsheets) that store the estimated "value" for each possible state or possible state-action pair in an environment.

+ The most common example is **Q-table**. This table has rows representing all possible states $(S)$ and columns representing all possible actions $(A)$. Each cell in the table, $Q(s, a)$, stores a numerical value that estimates the expected future reward for taking action 'a' in state 's' and then following an optimal policy thereafter. 

**Where were  they useful?**

Tabular methods are highly effective and can **guarantee** **convergence** to optimal policies in environment where:
+ The state space is small and discrete
+ The action space is small and discrete

**Limitations**:

The paper refers to this limitations as the "curse of dimensionality". This problem arises when the environment becomes more complex:

1. **Large state space**
2. **Continuous state spaces**
3. **Large action spaces**
4. **Storage constraints**
5. **Computation constraints**: Updating every cell in an enormous table during learning becomes computationally unfeasible.
6. **Lack of generalization**: Tabular methods treate each state-action pair idependantly. If an agent encounters a state it has never seen before, it has no way to estimate its vlaue because the cell in the table is empty. It can't generalize from similar past experience.


> DRL emerged by integrating DL with RL to overcome these limitations. DNNs are used as **function approximators** instead of explicit tables:

1. **Function Approximation**: Instead of storing a value for every *Q(s, a)* pair, DNN learns a function that can estimate *Q(s, a)* given state 's' and action 'a' as input. It's like learning the mathematicla formula to ccalculate the value, rather than **memorizing** every single result in a table.

2. **Handling Large and Continuous Spaces**:
    + **States**: A DNN can take high-dimensional or continuous state inputs (e.g., raw pixel data from an image, sensor readings).  The network's layer extract relevant feature and learn representations without needing to enumerate every possible state.
    + **Actions**: Depending on the DRL algorithm, DNNs can either output Q-values for a discrete set of actions or directly output continuos action value (e.g., a steering angle).
    + **Generalization**
    + **Reduced Storage and Computation**

#### Actor Critic method (policy based):

Actor-Critic algorithms combine two componenets to learn optimal **actions**:

1. **Actor(Policy)**: Decides what action to take in a given state. It learns a policy (a mapping from states to actions)
2. **Critic (Value Function)**: Evaluates how good the Actor's chosen action was by estimating the value of states or state-action pairs.

The Actor takes an action. The Critic observes the outcome and provides feedback(an "advantage" or "error" signal) to the Actor. The Actor uses this feedback to adjust and improve its policy (its decision-making process), while the Critic simultaneously refines its value estimates. This collabrative makes learning more efficient than relying on just one component.

**Use in Fog computing**:

Actor-Critic algorithms are used in fog computing for **computation offloading and resource allocation**:

+ **Problems**: IoT devices have limited capabilities and need efficient ways to decide where to process tasks(locally, fog, or cloud) while minimizing delay and energy consumption. Fog environments are dynamic and complex.
+ **Actor's Role**: Decides which task to offload and to which resource (e.g., specific fog node, cloud).
+ **Critic's Role**: Evaluates the chosen offloading decision based on metrics like latency and energy consumption. This feedback guides the Actor to make better future decisions.
+ A2C (Actor-Critic Advantage) and A3C(Asynchronous Advantage Actor-critic) are specific Actor-Critic algorithms used. They help minimize latency and optimize resource allocation in IoT-Fog-Cloud frameworks. Far instance, A2C has shown significant latency reductions.


**Difference between A2C and A3C**:

Both of them are Actor-Critic algorithms within Deep Reinforcement Learning, They share the core principle of having an "Actor" that decides on actions and a "Critic" that evaluates those actions. The primary difference lies in their **execution paradigm**, especially how they handle parallelization learning updates. 

**A3C (Asynchronous Advantage Actor-Critic):**

+ Asynchronous Nature: A3C operates by running multiple agents simultaneously and asynchronously in different copies of the environment. Each agent interacts with its own copy of the environment and calculates its own gradients (updates for the Actor and Critic networks).

+ Shared Global Network: These gradients are then asynchronously sent to a central, shared "global" neural network, which accumulates the updates.

+ Benefit: This asynchronous updating allows for diverse experiences from different agents to be used, which can stabilize learning and speed up the training process by overcoming issues like correlation in experiences. It also avoids the need for a replay buffer, which is common in other DRL algorithms like DQN.

**A2C (Advantage Actor-Critic):**

+ Synchronous Nature: A2C is essentially a synchronous version of A3C. Instead of agents updating the global network asynchronously, A2C waits for all parallel workers (or a batch of workers) to complete their experience collection and gradient calculations.

+ Batched Updates: Once all workers are done, their gradients are combined (averaged or summed) to perform a single, synchronous update to the global network.

+ Benefit: While A3C is known for its speed due to asynchronicity, A2C, with its synchronous updates, can sometimes lead to more stable and robust learning because the updates are based on a more consistent batch of experiences. It can also make more efficient use of hardware (like GPUs) by performing larger, batched computations.

### DQN (Deep Q-Network)

DQN is an off-policy Deep Reinforcement Learning algorithm that combines Q-learning with Deep Neural Networks (DNNs). It addresses the limitations of traditional Q-learning (tabular methods), specifically the "curse of dimensionality," which makes it infeasible to store and update Q-values in large or continuous state spaces.<br>
Instead of a Q-table, DQN uses a DNN to approximate the Q-function. The DNN takes a state as input and outputs the estimated Q-values for all possible actions in that state. It learns by minimizing the difference between its predicted Q-values and target Q-values, using techniques like experience replay and target networks to stabilize training.<br>

It's a popular technique for computation offloading problems in dynamic and complex environments like fog computing, aiming to minimize energy and latency.

### DRQN (Deep Recurrent Q-Network)

DRQN is an extension or variant of DQN. It incorporates Recurrent Neural Networks (RNNs), specifically Long Short-Term Memory (LSTM) units, into the DQN architecture.<br>

It's designed to handle problems in environments where the agent's observations are partially observable or when understanding the sequential nature and history of states is crucial for optimal decision-making. Standard DQN typically assumes full observability of the current state.<br>
By using recurrent layers (like LSTM), DRQN allows the network to maintain an internal "memory" of past observations. This memory helps the agent to infer the true state of the environment even when only partial information is available at any given moment, or to leverage historical context for better decision-making over time.<br>

In fog networks, DRQN has been applied to task offloading, particularly in SDN (Software-Defined Networking) architectures, to maximize processing tasks by satisfying delay constraints. It helps in scenarios where the task delay includes transmission latency, waiting delay, and processing delay, by choosing neighboring fog nodes to minimize latency.

### Dueling DDQN:

Dueling DDQN is an advanced variant of DQN that combines two key architectural improvements:
1. **Dueling Network Architecture**: This is the "Dueling" part. The neural network's final layer is split into two separate "streams" before combining them to produce the Q-value. One stream estimates the value of a state (V(s)), and the other estimates the advantage of each action in that state (A(s,a)). These two streams are then aggregated to produce the final Q-value.

2. **Double Q-learning(DDQN)**: This is the "Double" part. It addresses the overestimation bias that can occur in standard DQN. DDQN uses two separate Q-networks (a main network and a target network) to decouple the action selection from the action evaluation, leading to more accurate Q-value estimates.

The Dueling network effectively learns how good it is to be in a state independently of the actions, and then how much better (or worse) each action is compared to the average for that state. This allows the network to generalize better across actions. The Double Q-learning mechanism then uses one network to pick the best action and another (target) network to evaluate its Q-value, which stabilizes the learning process.

### DDPG (Deep Deterministic Policy Gradient):

DDPG is an off-policy Deep Reinforcement Learning algorithm that falls under the Actor-Critic framework. It's specifically designed for environments with continuous action spaces.

DDPG's main purpose is to enable RL agents to learn optimal policies in complex environments where actions are not discrete choices (like "move left" or "move right") but rather continuous values (like "apply 0.7 newtons of force" or "set steering angle to 15.3 degrees").

+ Deterministic Policy (Actor): Unlike some other Actor-Critic methods where the Actor outputs a probability distribution over actions, DDPG's Actor outputs a single, deterministic action for a given state. This is what makes it suitable for continuous action spaces.

+ Q-function (Critic): The Critic, similar to DQN, uses a Deep Neural Network to estimate the Q-value for a given state-action pair.

+ Target Networks & Experience Replay: Like DQN, DDPG employs target networks (separate, slower-updating copies of the Actor and Critic networks) and an experience replay buffer to stabilize learning and break correlations in observed experiences.

+ Learning Process: The Critic learns by minimizing the difference between its predicted Q-values and target Q-values. The Actor learns by using the Critic's Q-value to guide its policy, aiming to choose actions that maximize the Q-value
