### Jaccard Coefficient:

is a statistic used to measure the similarity between two finite sample sets. It quantifies how much two sets overlap.

The  formula for it:

$$J(A, B) = \frac{|A \cap B|}{|A \cup B|}$$

+ Where $|A \cap B|$ is the size(cardinality) of the intersection of sets A and B (the number of elements common to both sets).
+ $|A \cup B|$ is the size (cardinality) of the union of the sets A and B (The total number of unique elements present in either set).

The value of Jaccard Coefficient ranges from 0 to 1:

+ A value of 1 or 100 percent means the sets are **identical** 
+ A value of 0 means the sets have no common elements 
+ A value closer to 1 indicates greater similarity, while a value closer to 0 indicates greater dissimilarity.
