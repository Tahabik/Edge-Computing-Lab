## Contract Theory

Contract theory is a field in economics and law that studies how individuals and organizations construct contractual arrangements, typically in situations where information is asymmetrical (i.e., one party has more or better information than the other). It focuses on designing contracts that align the incentives of different parties to achieve a desired outcome, even when those parties have conflicting interests or private information.


**How it applies in the paper's context (vehicular fog computing):**
In reference [90], the authors proposed a resource management approach based on contract theory and Deep Q-Networks (DQN) for vehicular fog computing.

+ **Parties involved:**
    + **Service Providers (Fog Nodes/Cloud):** These entities have resources (computing, storage, networking) and want to offer them for task processing. They might have private information about their true capacity or cost of providing service.
    + **Service Consumers (Vehicles/IoT Devices):** These are the IoT devices or vehicles that need their tasks processed and want to achieve certain Quality of Service (QoS) levels (e.g., low latency, low energy consumption). They might have private information about the criticality or complexity of their tasks.

## Queuing Theory

Queuing theory is the mathematical study of waiting lines, or "queues." It analyzes processes involving arrivals, waiting, and service. It helps in understanding and optimizing systems where entities (customers, tasks, data packets) arrive, wait in a queue if resources are busy, and then get served.
