**Words around MINLP (mixed-integer non-linear programming)**:

It is a complex optimization problem formulation technique frequently encountered in journal papers that have multiple objectives and target multiple issues. Due to its high number of variables and constraints, MINLP problems are often classified as non-convex NP-hard problems.

Researchers commonly use MINLP to mathematically represent problems like:

+ Minimizing energy consumption under device latency constraints in MEC
+ Optimizing energy usage and latency problems in both signle-user and multiple-user MEC
+ Formulation energy usage and latency problems in both single-user and multiple-user MEC
+ Jointly optimizing resource allocation and computation resource scheduling to minimize energy consumption and computation time for mobile devices while meeting latency constraints.

Because of the high computational overhead and energy consumption associated with solving these complex problems optimally, researchers often seek alternative, approximation approaches to reduce complexity and make solutions applicable. These alternatives can include relaxation approaches, Lyapunov optimization, convex optimization, heuristic approaches, and machine learning algorithms, which aim to generate near-optimal results while accelerating convergence.


When discussing "relaxation" in the context of solving optimization problems like those in MEC, it refers to a technique used to simplify a difficult problem, typically one that NP-hard or non-convex. 


+ **Convex Relaxation**: This involves transforming a non-convex optimization problem (which is generally hard to solve) into a convex one. Convex problems have properties that make them much easier to solve efficiently, often guaranteeing a global optimum. This is frequently achieved by relaxing integer variables to be continuous (e.g., allowing a task to be 0.5 offloaded instead of just 0 or 1) or by approximating binary constraints with quadratic ones.

+ **Linear Relaxation**: A specific type of convex relaxation where integer variables (like 0 or 1 for offloading decisions) are relaxed to be continuous variables between 0 and 1. This turns a Mixed integer non-linear programming problem into a continuous non-linear programming problem, which can then be solved using techniques like linear programming or other continuous optimization methods.

**Signle-user MEC**: This scenario involves a single mobile device offloading its tasks to one or more MEC. The optimization problem in this case focuses on managing the offloading decisions and resource allocation for that one user to achieve objectives like minimizing energy consumption or latency.

**Multi-user MEC**: This scenario involves multiple mobile devices simultaneously offloading tasks to MEC servers. The complexity increases significantly here because the offloading decisions of one user can affect others, for example, through shared wireless channels or computational resources, leading to potential interference or resource contention. Optimization in multi-user MEC often aim to achieve system-wide goals (like total energy minimizatioin or overall latency reduction) while also considering fairness among users.

In the context of using RL and DRL for optimizing problems in MEC task offloading, here's a breakdown of the core components:

+ **Agent**: The "agent" is a decision-maker in RL system. In MEC offloading problems, the agent is typically the **mobile device** itself, or a **central controller/orchestrator**(e.g, an edge server or base station) that makes offloading decisions for multiple users. The agent interacts with the MEC environment to learn the best offloading policy.

+ **States**: The "state" represents the current situtation or observation of the MEC environment from the agent's perspective. It encompass all relevant information the agent needs to make an informed decision. For MEC offloading, common elements of a state can include:

    + **mobile device status**: Battery level, CPU utilization task queue length.
    + **Task characteristics**: Size of the task, computational requirements, deadline sensitivity(how much delay it can tolerate).
    + **MEC server status**: Available computational resources (CPU, memory), current load, processing capabilities, queue length at the edge server. 
    + **Wireless channel condition**: Available bandwidth, signal strength (e.g., SINR), interference levels, distance to the MEC server.
    + **Mobility information**: User's or vehicle's current location, speed, and predicted future movement (specially crutcial in VEC).

+ **Actionis**: An "action" is a decision or choice made by the agent at a given state. In MEC task offloading, actions typically relate to *where* and *how* to process a task. Example of actions:
    + **Local execution**: Process the task entirely on the mobile device
    + **Offloading to a specific MEC server**: send the task (or part of it) to a particular edge server for processing.
    + **Partial offloading**: Decide what proportion of a divisible task to offload versus executing locally.
    + **Resource allocation**: How much transmission power to use, or how much bandwidth to request for offloading.
    + **Task Scheduling**: The order in which tasks are processed or offloaded.

+ **Reward**: the "reward" is a numerical feedback signal the agent receives from the environments after taking an action in a given state.The agent's goal is to learn a policy that maximizes the cumulative reward over time. For MEC offloading, rewards are designed to reflect the optimization objectives of the problem:

    + Positive rewards: Might be given for successfully completing tasks quickly, saving energy, or achieving high QoS.
    + Negative rewards (penalties/costs): Are typically assigned for undesirable outcomes, such as
        + High latency/delay.
        + Excessive energy consumption.
        + Task failures or missed deadlines.
        + High computational overhead or network congestion.

In a game-theory-based solution, each "player" (e.g., a mobile device) makes decisions (e.g., whether to offload a task, how much energy to use) to maximize its own "utility" or "payoff" (e.g., minimizing its energy consumption or latency). We take problems into a MDP environment because MDPs provide a powerful mathematical framework for modeling decision-making in situations where outcomes are partly random and partly under the control of a decision-maker (agent). 

Why MDPs are particularly useful for problems like task offloading in MEC:
+ **Sequential Decision-Making**: Many real-world problems, especially in dynamic environments like MEC, involve a sequence of decisions over time rather than a single choice. An MEC offloading problem, for instance, might require continuous decisions about where to offload tasks as network conditions and device states change. MDPs are inherently designed for this kind of sequential decision-making.

+ **Uncertainty/Stochasticity**: The outcomes of actions in MEC environments are often uncertain. For example, network bandwidth can fluctuate, or a server's load can unpredictably change. MDPs explicitly incorporate this uncertainty through probabilities associated with state transitions.

+ **Goal-oriented learning**: In an MDP, the agent's goal is to maximize a cumulative numerical reward over time. This aligns perfectly with optimization problems in MEC where the objective is to, for instance, minimize latency and energy consumption (which can be framed as maximizing negative cost or reward).

+ **No Prior Model Needed (with RL)**: While MDPs define the problem structure, Reinforcement Learning (RL) algorithms (like Q-learning or DRL) are used to solve MDPs without needing a complete, pre-defined model of the environment's dynamics. This is crucial for complex MEC systems where explicitly modeling every possible state transition and reward might be impossible or computationally prohibitive. The agent learns the optimal policy directly through interaction and experience.

+ **Dynamic Environment**: MEC environments are highly dynamic due to mobile device movement, changing connectivity, and fluctuating computational resources. MDPs, particularly when combined with RL, provide a robust framework to learn and adapt to these changes over time.

what is the difference between FC (fog computing) and MEC (multi-access/mobile edge computing) paradigms?

While both FC and MEC aim to bring computing resources close to end-devices, there are some key differences:

+ **Management and Location of servers**: In MEC, offloading is managed by network operators because the servers are located within the Radio Access Network (RANs) of cellular systems. FC, on the other hand, provides computing, networking, storage, control, and acceleration from a remote cloud to user devices.

+ **Scope of Operation**: EC, a broader category that includes MEC, tends to operate only at the edge devices. FC, however, distributes storage, computing, networking, and control functions closer to the user device as a horizontal system-level architecture.

+ **Application and Markets**: FC was introduced to facilitate various areas of IoT and vertical markets like energy, industry, transportation, agriculture, and healthcare. MEC, meanwhile, has good flexibility for video analytics, location-based services, and augmented reality applications. MCC is more focused on providing remte computing resources for mobile applciations. 

+ **Capacity of Servers**: The capacity of FC and MEC servers is typically much smaller than that of CC

+ **Delay**: If the primary goal is to reduce delay, MEC is often considered the most desirable option.

How Blockchain helps in privacy and security?

Blockchain is a distributed, tamper-proof ledger maintained collectively by participating nodes. It provides:

**Data Integrity and tamper resistence**:
+ Every transaction (e.g, task offloading request, model update in FL, resource allocation) is stored in blocks. 
+ Each block is cryptographically linked (hashed) to the previous one. 
+ Once data is recorded, it cannot be altered without breaking the chain.
+ Ensures that no malicious node can modify task data or results undetected

**Decentralized Trust**:
+ Edge nodes and fog servers can interact without central trusted authority.
+ Consensus protocols (e.g., Proof of Stake, PBFT) validate transactions collectively.
+ Helps in secure coordination among untrusted devices.

**Privacy Protection**:
+ Blockchain and computation tasks are **logged immutably**
+ Any participant can audit **who accessed or modified what**.
+ Prevents data misuse and supports **accountability in edge services**.


**Smart Contracts for Automated Security Policies**:
+ Smart contracts = self-executing rules stored on blockchain 
+ Example: "Only authorized vehicles can offload tasks to this edge server."
+ Autonomous Authentication, payments, and task scheduling securely.

IRS -> Intelligent Reflecting Surface

An IRS is a programmable, passive or semi-passive metasurface that can intelligently control how electromagnetic waves propagate in a wireless environment. it is made up of many small reflecting elements, each of which can adjust:
+ phase
+ Amplitude
+ Polarization

Of the incident signals, allowing the network to reshape wireless channels without adding power-consuming elements like traditional base station or relays.

Normally wireless signals can be blocked, reflected randomly, or attenuated in complex invironments (e.g., urban areas, indoor networks)

In edge/computing networks, we often have:
+ Mobile users (IoT device, vehicles)
+ Edge servers (fog nodes)
+ Wireless communication links between them

An IRS can be deployed
+ On buildings or poles between user devices and fog nodes
+ To reflect and direct optimally, making wireless task offloading more reliable and energy efficient.

Example:

A device wants to offload a computation task to a fog node but has a weak direct link.<br>
IRS adjusts its reflection elements to maximize signal-to-noise ratio (SNR)<br>
This reduces required transmission power -> lower energy consumption for the device -> improves overall QoS in edge computing.
