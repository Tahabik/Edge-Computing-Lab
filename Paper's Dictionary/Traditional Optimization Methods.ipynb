## Non-linear Programming (NLP):

Non-linear programming is a mathematical **optimization** technique used to find the best possible solution (e.g., maximum profit, minimum cost, optimal allocation) for a problem where **one or more of the constraints or the objective function are nonlinear**.

+ **objective functioni**: A mathematical equation we want to maximize or minimize (e.g, total delay, energy consumption, profit).

+ **Constraints**: These are limitations or restrictions on the variables in our problem(e.g., limited processing power, network bandwidth, battery life).

+ **Nonlinear**: This is a key difference from linear programming. In a nonlinear problem, the relationships between variables in the objective function or the constraints are not straight lines when plotted on graph. They might involve curves, squares, logarithms, trigonometric function, or the complex relationships.

**Where was it useful(and its limitations as a "traditional method")?**

In the context of computation offloading in fog computing, nonlinear programming was used to try and solve problems like::
+ **Minimizing delay**: Deciding where to execute a task (locally, on a fog node, or in the cloud) to reduce the total time taken for completion.

+ **Minimizing Energy Consumption:** Optimizing task placement and execution to save battery life on IoT devices.

+ **Resource allocation**: Determining how to best assign limited computing, storage, and networking resources.

Papers states that these methods are good for simple environments and not suitable in problems and environments with dynamic and complex behaviours.

## Mixed Integer Programming(MIP)

Mixed Integer Programming is another mathematical optimization technique. It's an *extention* of linear programming (where all variables are continuous) and integer programming (Where all the varialbes **must** be integers). In MIP, some variables are restricted to be integers (e.g., 0 or 1, representing a "yes" or "no" decision, or a whole number representing a quantity), while others can be continuous (e.g., fraction of bandwidth).

Similar to NLP, MIP faces challenges in dynamic and complex fog environments:

+ Solving MIP problem is often NP-hard, meaning the time required to find an optimal solution grows exponentially with the problem size. For large scale fog network with many devices and tasks, finding the optimal solution in real-time becomes computationally prohibitive.

## Heuristic Algorithms

Heuristic algorithms are problem-solving techniques that employs a **practical** method **not guaranteed** to be optimal or perfect, but sufficient for reaching an **immediate**, **short-term goal**. They are often used when finding an **exact** or optimal solution is **impractical** or impossible due to the complexity or size of the problem. Instead of guaranteeing optimality, heuristics aim to find **good enough** solutions within a **reasonable amount of time**.

**How they were used to in computation  offloading**:

+ **Greedy offloading**: A simple heuristic might be to always offload a task to the nearest fog node or the one with the most available resources, without considering **long-term** consequences or global optimality.

+ **Rule-Based Systems**: Defining a set of rules, like "if latency is above X, offload; otherwise, execute locally."

+ **Load Balancing**: Distributing tasks among fog nodes based on simple criteria to prevent any single node from becoming overloaded.

While heuristics are faster and more adaptable than exact optimization methods like MIP for complex scenarios, they have significant drawbacks in dynamic fog computing.

+ **Sub-optimality**: By definition, heuristic don't guarantee the best possible solution. The "good enough" solution might be far from optimal, leading to inefficient resource usage, higher latency, or increased energy consumption compared to what's theoretically possible.
+ **Lack of adaptibility to Dynamics**: Although better than static mathematical programming, simple heuristic might not adapt intelligently to unforseen changes or complex interactions within the system. They often lack the "learning" component needed to truly optimize in an "uncertain and dynamic behavior like fog computing systems".
