## Lyapunov Optimization 

Lyapunov optimization is a mathematical framework used to design **stabilized control and scheduling algorithms** for **dynamic states** (especially in **networking**, **cloud computing**, and **queuing systems**) where the system evolves over time and decisions must be made sequentially.

It uses Lyapunnov functions, a concept from control theory, to ensure **stability** of a system also optimizing a performance objective (such as minimizing cost, delay, or energy consumption). 

+ A **Lyanpunov function** $L(t)$ is a scalar measure of the system's "congestion" or "energy" at time $t$. Example:

$$L(t) = \frac{1}{2} \sum_{i} Q_i(t)^2$$

where $Q_i(t)$ is the queue backlog for user $i$

+ The **Lapunov drift** measures how the function change over time:

$$\Delta(t) = E[L(t+1) - L(t)] \;\;\;\;\; | \;\;\;\;\; \text{system state at time t}$$

The goal is to keep this drift small or negative (to prevent instability).

+ **Drift-Plus-Penalty Framework**:

Instead of only minizing queue growth, the algorithm minimizes: 

$$\Delta (t) + V . Penalty(t)$$

Where:

+ $Penalty(t)$: cost function (e.g., energy use, delay, power consumption)
+ $V$: a control parameter that trades off between **queue stability** and **performance optimization**.

At each time slot $t$:
1. Observe the current system state (e.g, queue lengths, channel conditions, energy levels). 
2. Solve a simple optimization problem to minimize:

$$Drift(t) + V. Penalty(t)$$
3. Apply the decision (e.g., scheduling, power allocation, offloading task).
4. System evolves, repeat.
