### TSM: Temporal Segmentation and modules-based computation offloading using predictive analytics and NR-V2X


TSM reslies on status updates from previous time periods and usees predictive analytics to address the stochastic nature of vehicular networks.
This paper introduce a four-tier heirarchy of resource-rich nodes:

+ **Cloud-server**
+ **Local data center** 
+ **Road Side Unit(RSU)**
+ **Vehicular fog nodes(SVs)**


**Modular computation offloading structure that heuristically manages the whole process**

+ The vehicular Fog Computing-enabled scenarios, we can consider the **on-board computer unit(OBCU)** of its member smart automobiles as fog nodes, while the RSUs as edge nodes also known as a vehicular edge computing (VEC) server.


The main difference between an edge server and a fog node is their **distance** from a **device that needs computing resource(a "needy node")**.

+ **Fog Node**: A fog node is **closer** to the needy node. In a VFC system, then OBCUs of smart vehicles are considered fog nodes
+ **Edge Server**: An edge server is located at a relatively **greater distance** from the needy node than a fog node. In the VFC scenario, RSUs are considered edge nodes, also known as VEC servers.

**More about TSM**

It is a new method for **computation offloading** in VFC env. It's designed specifically for **delay-sensitive, computation-intensive** vehicular applications that needs to be processed quickly.

This method works in short, discrete time periods. It addresses the unpredictable nature of vehicular networks by using **predictive analytics** based on status updates from previous time periods. This approach relies on a four-tier hierarchy of resource-rich nodes, which includes smart vehicles, RSUs, data centers, cloud servers.

The process is managed **heuristically** through an online, modular computation offloading structure:
+ **Module 1: First-phase decision for task offloading**:
    + A potential controller node (which can be an RSU or a vehicular fog node) receives a task request from a needy node
    + It uses Algorithm 1 to decide whether to offload the task itself or to one of four potential donor nodes based on resource availability and an **estimation** total response time 

+ **Module 2: The second phase of intra-fog-vehicle(SV) scheduling**:
    + If the first module decides to offload to a vehicular fog node(SV), Algorithm 2 is invoked to select the most suitable one.
    + The algorithm sorts tasks by computational poweor (descending) and available vehicles by computational power(ascending) to efficiently pair high-demand tasks with the lowest-resource-requirement vehicles.

**NR-V2X guidelines' Side Link**: This refers to a specific type of direct communication between devices using 5G New Radio (NR) standard, as defined by the 3rd Generation Parternship Project (3GPP):

+ **Side Link**: It is a technology that allows for direct-to-device (D2D) communication without having to route the data through a cellular base station (gNB). This is crucial for V2X(Vehicle-to-Everything) applications, as it enables low-latency communication between vehicles, pedestrians, and other road users.
+ **NR-V2X**: This is the latest evolution of cellular V2X(C-V2X) communication, based on 5G NR technology. It is designed to support advanced V2X services with stricter Quality of Service requirements, such as ultra-low latency and high reliability, that older standards like LTE-V2X cannot meet.
+ **Modes of sidelink**: The document mentions that the TSM framework employs the guidelines of all sub-modes of NR-V2X's Mode-2. The modes define how resources are allocated for sidelink communication:
    + **Mode-1**: This is a centralized allocation mode where a cellular base station (gNB) schedules the sidelink resources to be used by the User Equipment (UE). The paper mentions this mode is used for cellular links like V2I, V2N, and P2I.
    + **Mode-2**: In this mode, the UE (such as a vehicle or a pedestrian device) autonomously determines which sidelink transmission resources to use. The paper states that this mode is used for ad-hoc links like V2V and V2P, as these interactions don't depend on a gNB for task scheduling and allocation.
